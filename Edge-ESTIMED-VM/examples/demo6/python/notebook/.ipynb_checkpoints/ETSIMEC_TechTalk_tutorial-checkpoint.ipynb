## (2) Initialising the MEC App using the MEC Application Support API (MEC011)

For this part of the tutorial, we will use the MEC Application Support API ```mec_app_support```, version 2.

````python
mec011_app_support_path = mec_base_path + "/mec_app_support/v2"
````

### Notify the MEP that this instance is up and running
- The confirm_ready method is used by the MEC application instance to notify the MEC platform that it is up and running.





````python
url =  mec011_app_support_path + "/applications/" + appInstanceId  + "/confirm_ready"
payload_dic = {"indication": "READY"}   # <-- Message content
method = "POST"        # <-- GET, POST, DELETE

response =  requests.request(
    method, 
    url, 
    data=json.dumps(payload_dic), 
    headers={'accept': 'application/json'}, 
    params={})

print("Status Code", response.status_code)
````

### Check  Subscriptions

The GET method may be used to request information about all subscriptions for this requestor. Upon success, the response contains message content with all the subscriptions for the requestor.


[swagger](https://forge.etsi.org/swagger/ui/?url=https://forge.etsi.org/rep/mec/gs011-app-enablement-api/-/raw/master/MecAppSupportApi.yaml#/appSubscriptions/ApplicationsSubscriptions_GET)

````python
url = mec011_app_support_path + "/applications/" + appInstanceId  + "/subscriptions"
payload_dic = {}
method = "GET"  # <-- GET, POST, DELETE

response =  requests.request(
    method, 
    url , 
    data=json.dumps(payload_dic), 
    headers={'accept': 'application/json'}, 
    params={})      

print("Status Code", response.status_code) 
````

### Subscribe to AppTermination Notification
The POST method may be used to create a new subscription. One example use case is to create a new subscription to the MEC service availability notifications. Upon success, the response contains entity body describing the created subscription.


[swagger](https://forge.etsi.org/swagger/ui/?url=https://forge.etsi.org/rep/mec/gs011-app-enablement-api/-/raw/master/MecAppSupportApi.yaml#/appSubscriptions/ApplicationsSubscriptions_POST)


````python
url = mec011_app_support_path + "/applications/" + appInstanceId  + "/subscriptions"
payload_dic = {
  "subscriptionType": "AppTerminationNotificationSubscription",
  "callbackReference": "http://my.callback.com/mec_service_mgmt_ser_availabilities/terminate",    
  "appInstanceId": appInstanceId
}
method = "POST"  # <-- GET, POST, DELETE

response =  requests.request(
    method, 
    url, 
    data=json.dumps(payload_dic), 
    headers={'accept': 'application/json'}, 
    params={})

print("Status Code", response.status_code)
````

### Check  Subscriptions

The GET method may be used to request information about all subscriptions for this requestor. Upon success, the response contains message content with all the subscriptions for the requestor.


[swagger](https://forge.etsi.org/swagger/ui/?url=https://forge.etsi.org/rep/mec/gs011-app-enablement-api/-/raw/master/MecAppSupportApi.yaml#/appSubscriptions/ApplicationsSubscriptions_GET)


````python
url = mec011_app_support_path + "/applications/" + appInstanceId  + "/subscriptions"
payload_dic = {}
method = "GET"  # <-- GET, POST, DELETE

response =  requests.request(
    method, 
    url , 
    data=json.dumps(payload_dic), 
    headers={'accept': 'application/json'}, 
    params={})      

print("Status Code", response.status_code) 
````

### Check if the Location Services are available in the choosen MEP 
For this part of the tutorial, we will use theWe will need to use the MEC Service Management (MEC011) API ```mec_service_mgmt```, version 1:

````python
mec011_service_mgmt_path = mec_base_path + "/mec_service_mgmt/v1"
````

The following method retrieves information about a list of mecService resources. This method is typically used in "service availability query" procedure


[swagger](https://forge.etsi.org/swagger/ui/?url=https://forge.etsi.org/rep/mec/gs011-app-enablement-api/-/raw/master/MecServiceMgmtApi.yaml#/services/Services_GET)

###  List of  zones 

To pick a zone of interest (e.g. zone with the most terminals) lets first check the list the avaliable ones usin the Location APIs. This query retrieves information about one or more specific zones or a list of zones. The output is a list of identifiers for zones authorized for use by the application. 

[swagger](https://forge.etsi.org/swagger/ui/?url=https://forge.etsi.org/rep/mec/gs013-location-api/raw/master/LocationAPI.yaml#/location/zonesGET)



````python
url = mec013_path + "/queries/zones"
payload_dic = {}
method = "GET"

response =  requests.request(
    method, 
    url, 
    data=json.dumps(payload_dic), 
    headers={'accept': 'application/json'}, 
    params={})

print("Status Code", response.status_code)
````

````python
zone_list = response.json()['zoneList']['zone']
pd.DataFrame.from_dict(zone_list).reindex(columns=["zoneId", "numberOfUsers", "numberOfAccessPoints", "numberOfUnserviceableAccessPoints"]).style
````

### Subscribe zoneLocationEventSubscription (notifies every time there is a zone change)

Subscribe to the service that notifies everytime there is change in a zone.


[swagger](https://forge.etsi.org/swagger/ui/?url=https://forge.etsi.org/rep/mec/gs013-location-api/raw/master/LocationAPI.yaml#/location/zoneSubPOST)


````python
url = mec013_path + "/subscriptions/zones"
payload_dic = {}
method = "POST"

payload_dic ={
  "zoneLocationEventSubscription": {
    "subscriptionType": "ZoneLocationEventSubscription",
      # URI exposed by the client on which to receive notifications via HTTP
    "callbackReference": "http://my.callback.com/zone-location-event-notification/zone3", 
      # Identifier of zone (e.g. zone001) to monitor.
    "zoneId": "zone03",
      # A correlator that the client can use to tag this particular resource representation during a request to create a resource on the server. 
    "clientCorrelator": "0123",
    "requestTestNotification": True,
       # List of the users to be monitored. If not present, all the users need to be monitored.
    "addressList": [
      "10.1.0.1",
      "10.100.0.1",
      "10.10.0.1",
      "10.100.0.2",
      "10.100.0.3"
    ],
    "reportingCtrl": {
        # Maximum number of notifications. Default is 0 (no max value)
      "maximumCount": 150,
        # Maximum frequency (in seconds) of notifications per subscription.
      "maximumFrequency": 0,
        # Minimum interval between reports in case frequently reporting. Unit is second.
      "minimumInterval": 5
    },
      # List of user event values to generate notifications for.
    "locationEventCriteria": [
      "ENTERING_AREA_EVENT", "LEAVING_AREA_EVENT"
    ],
    "expiryDeadline": {
      "seconds": 1977836800,
      "nanoseconds": 0
    }
  }
}

response =  requests.request(
    method, 
    url, 
    data=json.dumps(payload_dic), 
    headers={'accept': 'application/json'}, 
    params={})

print("Status Code", response.status_code)
````

## (4) Gracefully terminating a MEC Application



<div class="alert alert-block alert-info">
When the MEC application instance is notified to gracefully terminate, it provides to the MEC platform 
that the application has completed its application level related terminate/stop action</div>



###  Delete the userLocationEventSubscription
The subscription DELETE method in MEC013 is used to cancel the existing ```zoneLocationEventSubscription```.





````python
# GET subscriptions list
url = mec013_path + "/subscriptions/zones"
payload_dic = {}
method = "GET"

response =  requests.request(
    method, url , 
    data=json.dumps(payload_dic), 
    headers={'accept': 'application/json'}, 
    params={})

print("Status Code", response.status_code)
````

````python
subscription_list = response.json()['notificationSubscriptionList']['subscription']
len(subscription_list)
````

````python
payload_dic = {}
method = "DELETE"

for subscription in response.json()['notificationSubscriptionList']['subscription']:
    url = subscription['href']
    response =  requests.request(
        method, url , 
        data=json.dumps(payload_dic), 
        headers={'accept': 'application/json'}, 
        params={})
    print("Status Code", response.status_code)
````

### Delete the AppTermination subscription

The following method in the MEC011 Application Support can be used to delete the ```AppTerminationNotificationSubscription```.



But first ywe need to retrieve the subscription ID:




````python
url = mec011_app_support_path + "/applications/" + appInstanceId  + "/subscriptions"
payload_dic = {}
method = "GET"

response =  requests.request(
    method, url , 
    data=json.dumps(payload_dic), 
    headers={'accept': 'application/json'}, 
    params={})      

print("Status Code", response.status_code) 
````

````python
len(response.json()['_links']['subscriptions'])
````

````python
import re
for subscription in response.json()['_links']['subscriptions']:
    print("Subscription Type " +  subscription['subscriptionType'] + " " +
          "Subscription ID " + re.findall('/([\w\-\.]+)', subscription['href'])[8])
````

````python
app_subscription_list = response.json()['_links']['subscriptions']
payload_dic = {}
method = "DELETE"

for subscription in app_subscription_list:
    response =  requests.request(
        method, 
        subscription['href'] , 
        data=json.dumps(payload_dic), 
        headers={'accept': 'application/json'}, 
        params={})      

    print("Subscription Type " +  subscription['subscriptionType'] + " " +
          "Subscription ID " + re.findall('/([\w\-\.]+)', subscription['href'])[8])
    print("Status Code", response.status_code) 
````
